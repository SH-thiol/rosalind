# Fn = Fn−1 + Fn−2    Recursive definition of the Fibonacci sequence. 
체감 난이도 높음. 

```python
def fib(n, k):   # 함수 정의 (변수명); 토끼가 한 쌍이 태어나서 매달 새끼를 k쌍씩, n개월동안 낳는다.
    if n <=2:    # ← 콜론 필수
        return 1 # ← 들여쓰기 필수 (스페이스 4칸 또는 탭)
    else:        # ← 콜론 필수 (else는 선택사항이지만, 학습초반에는 명시적으로 사용하는 것이 좋다.)
        return fib(n-1) + 3*fib(n-2) # ← 들여쓰기 필수 ###수식오류!

fib(5, 3)
```

```python
def fib(n, k):
    if n <=2:
        return 1
    return fib(n-1, k) + k*fib(n-2, k) # 재귀함수, 고정된 k값을 계속 사용하기 위해서 매개변수로 k를 추가해준다.

fib(5, 3)
```

```python
def fib(n, k):
    if n <=2:
        return 1
    return fib(n-1, k) + k*fib(n-2, k)

fib(28, 5)
```
