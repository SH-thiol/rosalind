```python
from Bio import SeqIO
from Bio.Seq import Seq

with open("cons_sample.txt", "r") as f:
    for rosalind in SeqIO.parse(f, "fasta"):
        print(rosalind.id)
        print(rosalind.seq) # 마지막 서열만 출력됨
    # for문이 끝나면 rosalind 변수에는 마지막 서열 정보가 저장되어 있기 때문
```

```python
seqs = []
with open("cons_sample.txt", "r") as f:
    for rosalind in SeqIO.parse(f, "fasta"):
        seqs.append(str(rosalind.seq)) # 문자열로 변환하여 리스트에 추가

print(seqs)           # 모든 서열 확인
print(zip(*seqs))       # list를 사용하지 않으면 zip 객체(object)로 메모리 주소만 보이게 출력됨
print(list(zip(*seqs)))  # *를 사용하여 각 위치의 염기들을 묶어서 확인

DNA_strings = seqs
print(DNA_strings)
```

```python
from collections import Counter 
#Counter는 collections 모듈에서 제공하는 클래스 중 하나로, 주어진 데이터에서 각 요소의 빈도수를 계산하는 데 사용됨

DNA_strings = seqs
consensus = ""
profile = {base: [] for base in "ACGT"}
###profile = {"A": [], "C": [], "G": [], "T": []}

for pairs in zip(*DNA_strings):

    count = Counter(pairs)
    print(count)  # 각 위치의 염기 빈도수 확인

    consensus += count.most_common(1)[0][0]  # 가장 빈도가 높은 염기를 consensus에 추가
    #most_common(1)은 가장 빈도가 높은 요소를 하나 반환하는 메서드로, [0][0]을 통해 염기 문자만 추출
    #count.most_common(1) -> [('A', 5)]
    #count.most_common(1)[0] -> ('A', 5)
    #count.most_common(1)[0][0] -> 'A'

    for base in "ACGT":
        profile[base].append(count[base])  # 각 염기의 빈도수를 profile에 추가  
    
print(profile)  # profile 딕셔너리 확인

for base in "ACGT":    
    print(f"{base}: {' '.join(map(str, profile[base]))}") # 각 염기의 빈도수를 문자열로 변환하여 출력
#map(str, profile[base])는 profile[base] 리스트의 각 요소를 문자열로 변환하는 함수
#profile[base]는 딕셔너리에서 base 키에 해당하는 리스트를 가져옴. 예를 들어, profile['A']는 'A' 염기의 빈도수 리스트

print(consensus)  # 최종 consensus 문자열 출력
```

```python
seqs = []
with open("rosalind_cons.txt", "r") as f:
    for rosalind in SeqIO.parse(f, "fasta"):
        seqs.append(str(rosalind.seq)) # 문자열로 변환하여 리스트에 추가

from collections import Counter 

DNA_strings = seqs
consensus = ""
profile = {base: [] for base in "ACGT"}

for pairs in zip(*DNA_strings):
    count = Counter(pairs)
    consensus += count.most_common(1)[0][0]

    for base in "ACGT":
        profile[base].append(count[base])  

print(consensus)
for base in "ACGT":    
    print(f"{base}: {' '.join(map(str, profile[base]))}") 


```
