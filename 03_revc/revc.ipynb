```python
table = str.maketrans("ATGC", "TACG")

with open("rosalind_revc.txt", "r") as f:
    dna = f.read().strip()    
revc = dna.translate(table)

print(revc[::-1]) #[::-1]은 reverse 방법, [start:stop:step], step이 -1이면 뒤에서부터 1칸씩 이동하며 출력
```
