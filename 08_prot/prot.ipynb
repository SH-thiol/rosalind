```python
s = "AUGGCCAUGGCGCCCAGAACUGAGAUCAAUAGUACCCGUAUUAACGGGUGA"
for i in range(0, len(s), 3):
    codon = s[i:i+3]
    if codon == "AUG":
        print("Start codon found at position", i)
```

```python
from Bio.Seq import Seq
# from Bio import SeqIO --- Bio 안의 SeqIO 모듈 가져오기
# from Bio.Seq import Seq --- Bio 안의 Seq 모듈(폴더) 안의 Seq 클래스 가져오기

rna_string = "AUGGCCAUGGCGCCCAGAACUGAGAUCAAUAGUACCCGUAUUAACGGGUGA"
rna_sequence = Seq(rna_string)
protein = rna_sequence.translate()
print(protein)
print(protein[:-1])  # 마지막 문자 제거 (종료 코돈*)
```

```python
from Bio.Seq import Seq
with open ("rosalind_prot.txt", "r") as f:
    rna = f.read().strip()
    print(rna)
    print("The number of amino acids = ", len(rna)//3)

    rna_sequence = Seq(rna) # RNA 문자열을 Seq 객체로 변환
    prot = rna_sequence.translate() # RNA 시퀀스를 단백질로 번역
    print("The number of prot+stop codon = ", len(prot))
    print(prot[:-1])
```
