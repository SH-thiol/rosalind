```python
from Bio.Seq import Seq
list_of_seqs = [Seq("ACGT"), Seq("AACC"), Seq("GGTT")]
concatenated = Seq("")
for s in list_of_seqs: #add(+)나 .join도 사용 가능, for loop like this:
    concatenated += s
    # return concatenated(X) return 은 함수 안에서만 쓸 수 있음

print(concatenated)
```
