```python
from Bio.Seq import Seq

s = Seq("GATATATGCATATACTT")
t = Seq("ATAT")

s.index(t) # 문자열에서 t가 처음 나오는 위치 반환, 없으면 ValueError 발생
# 09번에서 정의한 positions는 index에 +1을 해주어야 함
```

```python
#방법 1. index에 +1
position = [i for i in range(len(s)) if s[i:i+len(t)] == t] 
print(position)
# 문자열 s에서 t와 일치하는 부분의 시작 위치를 찾는 방법
#i에 +1을 해주는 이유는 문자열의 인덱스가 0부터 시작하기 때문임.
position = [i + 1 for i in range(len(s)) if s[i:i+len(t)] == t] 
print(position)

#1. 문자열 s의 각 위치 i에 대해, s[i:i+len(t)]가 t("ATAT")와 일치하는지 확인 (0:4, 1:5, 2:6, ...)
#2. 일치하는 경우, 해당 위치 i를 결과 리스트에 추가 (position.append(i + 1))
#3. 최종적으로, 결과 리스트에는 t가 s에서 일치하는 모든 위치가 포함됨
```

```python
#방법 2. find() 사용
positions = [] # t가 s에서 일치하는 위치를 저장할 빈 리스트
start = 0 # 검색을 시작할 위치, 처음에는 0부터 시작
while True: # 무한 루프 시작
    i = s.find(t, start) # 문자열 s에서 t를 start 위치부터 찾음, 없으면 -1 반환
    if i == -1:
        break # 더 이상 t가 s에 없으면 반복 종료
    positions.append(i + 1) # 찾은 위치 i에 +1을 해서 positions 리스트에 추가
    start = i + 1 # 다음 검색은 i + 1 위치부터 시작 (중복된 부분도 찾기 위해)

print(positions)
```

```python
with open("rosalind_subs.txt") as f:
    a = Seq(f.readline().strip())
    b = Seq(f.readline().strip())
print(len(a), len(b))

repeat_positions = [] 
start = 0 
while True: 
    i = a.find(b, start)
    if i == -1:
        break 
    repeat_positions.append(i + 1)
    start = i + 1

print(repeat_positions) 
print(' '.join(map(str, repeat_positions))) # 리스트의 정수들을 문자열로 변환하여 공백으로 구분하여 출력
print(''.join(map(str, repeat_positions)))
```
