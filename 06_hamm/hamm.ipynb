```python
s = "GAGCCTACTAACGGGAT"
t = "CATCGTAATGACGGCCT"

print(s == t)

def hamming_distance(s, t): 
    pairs = zip(s, t) #zip() 함수를 사용하여 s와 t의 각 문자 한개씩 쌍(a, b)으로 묶음
    distance = 0 #거리(distance)를 0으로 초기화해서 변수 distance에 저장
    for a, b in pairs:  #s, t랑 중복되지 않는 변수명으로 a, b를 사용
        if a != b: #a와 b가 다르면
            distance += 1 #distance에 1을 더해서 누적
    return distance #return print(distance)로 바꾸면 함수가 None을 반환, distance값이 출력되지 않음. 
    #return은 함수의 결과값을 반환하는 것, print는 화면에 값을 출력하는 것. 

print(hamming_distance(s, t)) #값을 출력해서 볼 수 있도록 해주어야 함
```

```python

def hamming_distance(u, v): 
    pairs = zip(u, v)
    distance = 0
    for a, b in pairs:
        if a != b:
            distance += 1
    return distance

print(hamming_distance(u, v))
```
