```python
k = 2  #k = individuals are homozygous dominant for a factor, ; AA
m = 2  #m are heterozygous recessive, ; Aa
n = 2  #n are homozygous recessive ; aa


total = k + m + n
print(total)


```

```python
def C(n, r):
    return n*(n-1) / 2 #곱셈연산자를 빠뜨리면 안됨
print(C(total, 2)) 
#Aa & Aa = 1/15, Aa & aa = 4/15, aa & aa = 1/15
#AaAa에서 aa가 나올 확률은 1/4, Aaaa에서 aa가 나올 확률은 1/2, aaaa에서 aa가 나올 확률은 1

print(1-(1/15 * 1/4 + 4/15 * 1/2 + 1/15 * 1))
```

```python
from math import comb #Python에서 조합(combination)을 계산하는 함수
total = k + m + n
total_combination = (comb(total, 2))
print(total)
print(total_combination)

C_AaAa = comb(m, 2) / total_combination #Aa와 Aa가 짝을 이루는 경우의 수를 계산_조합함수
C_Aaaa = (m*n)/total_combination #Aa와 aa가 짝을 이루는 경우의 수를 계산_곱셈연산자
C_aaaa = (comb(n, 2) / total_combination) #aa와 aa가 짝을 이루는 경우의 수를 계산_조합함수

Aa_aa_one = (m+n)/total_combination #Aa와 aa 중에 아무거나 하나만 선택하는 경우의 수를 계산

print(1 - (C_AaAa * 1/4 + C_Aaaa * 1/2 + C_aaaa * 1))
    
```
