```python
from Bio import SeqIO # Biopython 라이브러리(공구함)의 SeqIO 모듈(서랍) 안의 SeqIO.parse() 함수(도구) 
for record in SeqIO.parse("sample.fasta", "fasta"): #파일명(확장자까지 써주어야), 포맷명(어떤 포맷을 읽는지)
    #SeqIO.parse() 는 FASTA 말고 다른 형식도 읽을 수 있음, 파일 이름으로 Python이 자동으로 형식을 판단하지 못함
    print(record.id)
    print(record.seq) # record.seq는 Seq 객체로, 문자열과 유사하지만 문자열이 아님. 
                      #문자열로 변환하려면 str(record.seq) 사용.
    
```

```python
def gc_content(seq):
    gc_count = seq.count("G") + seq.count("C")
    return gc_count / len(seq) * 100
```

```python
print(gc_content("Rosalind_0808")) ### SeqIO로 읽어 모듈 DNA 서열을 넣어야함. 
```

```python
for record in SeqIO.parse("sample.fasta", "fasta"): 
    print(record.id) # ← 들여쓰기 필수 (for문 안에 있어야 함)
    print(gc_content(record.seq)) # ← 들여쓰기 필수 (for문 안에 있어야 함)
```

```python
# max()함수를 적용할 때는, for문을 사용하지 않고, key 매개변수에 lambda 함수를 사용

records = list(SeqIO.parse("sample.fasta", "fasta")) #records 리스트 [rosalind_xxx1, rosalind_xxx2,...]
max_record = max(records, key=lambda r: gc_content(r.seq)) 
    # max() 함수는 records 리스트에서 gc_content(r.seq)가 가장 큰 레코드를 반환
    # key=무엇을 기준으로 비교할지, lambda r:r이라는 임시 변수를 받아서, gc_content(r.seq)GC content를 계산한 값으로 비교
    #lambda는 익명 함수, 간단한 함수를 한 줄로 작성할 때 사용

print(f"{max_record.id}")
print(f"{gc_content(max_record.seq):.4f}") # :.4f는 f-string에서 소수점 4자리까지 출력하라는 의미
```
